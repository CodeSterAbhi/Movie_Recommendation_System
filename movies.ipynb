```python
# so, genres contain a dictionary of id and name.. but we only need the names from it
import ast
df['genres'] = df['genres'].apply(lambda x: " ".join(i['name'] for i in ast.literal_eval(x)))
```

```python
# No more null values means we are good to go for the combining the columns
df['tags'] = df['overview'] + " " + df["genres"] + " " + df["tagline"]
```

## Vector embedding using NLTK

```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import re
```

```python
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
```

```python
# function for preprocessing the text
def preprocess_text(text):
  # Lowercase
  text = str(text).lower()

  # remove puctuations
  text = re.sub(r'[^\w\s]', '', text)

  # remove stopwords
  words = text.split()   # first we need to split the text into words
  words = [word for word in words if word not in stop_words]

  # Lemmatize the words
  words = [lemmatizer.lemmatize(word) for word in words]

  return " ".join(words)
```

```python
df['tags'] = df['tags'].apply(preprocess_text)
```

```python
# Lets reset the index
df = df.reset_index(drop=True)
```

```python
# Make a indices which can be used for slicing of titles later
indices = pd.Series(df.index, index=df['title']).drop_duplicates()
```

## TF-IDF vectorize

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer(max_features=50000, ngram_range=(1,2), stop_words='english')
```

```python
tfidf_matrix = tfidf.fit_transform(df['tags'])
```

```python
tfidf_matrix
```

## Cosine similarity

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
def recommend(title, n=10):
  if title not in indices:
    return ['Movie Not Found']

  idx = indices[title]
  sim_score = cosine_similarity(tfidf_matrix[idx], tfidf_matrix).flatten()
  similar_idx = sim_score.argsort()[::-1][1:n+1]
  return df['title'].iloc[similar_idx]
```

```python
recommend('Avatar', 5)
```
